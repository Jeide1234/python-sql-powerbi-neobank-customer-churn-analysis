```python
import pandas as pd
```

```python
import numpy as np
```

```python
# Load the dataset
df = pd.read_excel("Project_2_Neobank_Churn_Engagement_Data.xlsx")

# Standardise date columns
df["Signup_Date"] = pd.to_datetime(
    df["Signup_Date"],
    errors="coerce",
    dayfirst=True
)

df["Snapshot_Month"] = pd.to_datetime(
    df["Snapshot_Month"],
    errors="coerce"
)

# Enforce numeric data types
numeric_columns = [
    "Account_Balance",
    "Monthly_Deposits",
    "Monthly_Transactions",
    "App_Sessions_30D",
    "Failed_Logins",
    "Support_Tickets",
    "Core_Feature_Score",
    "Engagement_Rate",
    "AI_Churn_Risk_Score"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )

# Check whether type conversion created any missing values
print("Missing values after data type conversion:")
print(df[numeric_columns].isnull().sum())
```

```python
# Check the size of the dataset
print("Dataset Shape:")
print(df.shape)

print("\n")

# Display column names
print("Column Names:")
print(df.columns)

print("\n")

# Check data types
print("Data Types:")
print(df.dtypes)
```

```python
# Check for missing values
print("Missing Values:")
print(df.isnull().sum())

print("\n")

# Check for duplicate records
print("Duplicate Rows:")
print(df.duplicated().sum())
```

```python
# Detect impossible customer behaviour
anomalies = df[
    (df["Monthly_Transactions"] > 0) &
    (df["App_Sessions_30D"] == 0)
]

print("Number of Anomalies:")
print(anomalies.shape[0])

anomalies.head()
```

```python
# Save anomalous records
anomalies.to_csv("quarantined_churn_data.csv", index=False)

print("Quarantine file created successfully.")
```

```python
# Remove anomalies from the main dataset
df_clean = df[
    ~(
        (df["Monthly_Transactions"] > 0) &
        (df["App_Sessions_30D"] == 0)
    )
].copy()

print("Original Dataset:", df.shape)
print("Clean Dataset:", df_clean.shape)
```

```python
# Recalculate AI Risk Band
#
# Assumption:
# The project brief does not specify the Core_Feature_Score thresholds
# for assigning AI_Risk_Band. Therefore, the following business rules
# are assumed:
#
# Core_Feature_Score < 50   = High Risk
# Core_Feature_Score 50–79  = Medium Risk
# Core_Feature_Score >= 80  = Low Risk

conditions = [
    df_clean["Core_Feature_Score"] < 50,
    (df_clean["Core_Feature_Score"] >= 50) &
    (df_clean["Core_Feature_Score"] < 80),
    df_clean["Core_Feature_Score"] >= 80
]

choices = ["High", "Medium", "Low"]

df_clean["AI_Risk_Band"] = np.select(
    conditions,
    choices,
    default="Medium"
)

# Check the results
df_clean["AI_Risk_Band"].value_counts()
```

```python
# Customer Friction Score

df_clean["Customer_Friction_Score"] = (
    100
    + np.where(df_clean["Plan_Type"] == "Premium", 10, 0)
    - (25 * df_clean["Failed_Logins"])
    - np.where(df_clean["KYC_Status"] == "Pending", 40, 0)
    - (5 * df_clean["Support_Tickets"])
)

# Ensure score doesn't go below 0
df_clean["Customer_Friction_Score"] = (
    df_clean["Customer_Friction_Score"]
    .clip(lower=0)
)

# View the results
df_clean[
    [
        "Customer_ID",
        "Plan_Type",
        "KYC_Status",
        "Failed_Logins",
        "Support_Tickets",
        "Customer_Friction_Score"
    ]
].head()
```

```python
# Calculate transaction volatility by plan type

plan_volatility = (
    df_clean.groupby("Plan_Type")["Monthly_Transactions"]
    .agg(["mean", "std"])
)

plan_volatility["Plan_Transaction_Volatility_Index"] = (
    plan_volatility["std"] / plan_volatility["mean"]
)

plan_volatility
```

```python
# Map the volatility index back to the main dataset

df_clean["Plan_Transaction_Volatility_Index"] = (
    df_clean["Plan_Type"].map(
        plan_volatility["Plan_Transaction_Volatility_Index"]
    )
)

df_clean[
    [
        "Customer_ID",
        "Plan_Type",
        "Monthly_Transactions",
        "Plan_Transaction_Volatility_Index"
    ]
].head()
```

```python
# Export the final enriched dataset

df_clean.to_csv(
    "neobank_customer_churn_cleaned.csv",
    index=False
)

print("Cleaned dataset exported successfully.")
```

```python
import urllib.parse
from sqlalchemy import create_engine, text

server = r"localhost\MSSQLSERVER02"
database = "Neobank_Analytics"

connection_string = urllib.parse.quote_plus(
    "DRIVER={ODBC Driver 17 for SQL Server};"
    f"SERVER={server};"
    f"DATABASE={database};"
    "Trusted_Connection=yes;"
    "TrustServerCertificate=yes;"
)

engine = create_engine(
    f"mssql+pyodbc:///?odbc_connect={connection_string}",
    fast_executemany=True
)

print("SQL Server connection engine created successfully.")
```

```python
df_clean.to_sql(
    name="neobank_customer_churn_pipeline_test",
    con=engine,
    schema="dbo",
    if_exists="replace",
    index=False,
    chunksize=500
)

print("Test pipeline completed successfully.")
print(f"{len(df_clean)} rows loaded into the SQL Server test table.")
```

```python
with engine.connect() as connection:
    total_rows = connection.execute(
        text("""
        SELECT COUNT(*)
        FROM dbo.neobank_customer_churn_pipeline_test
        """)
    ).scalar()

print("Rows confirmed in SQL Server:", total_rows)
```

```python
from sqlalchemy import inspect

production_table = "neobank_customer_churn_cleaned"
schema_name = "dbo"

# Read the existing SQL Server table structure
inspector = inspect(engine)

sql_columns = [
    column["name"]
    for column in inspector.get_columns(
        production_table,
        schema=schema_name
    )
]

python_columns = df_clean.columns.tolist()

print("Python columns:", len(python_columns))
print("SQL columns:", len(sql_columns))

if python_columns != sql_columns:
    missing_in_sql = [
        column for column in python_columns
        if column not in sql_columns
    ]

    missing_in_python = [
        column for column in sql_columns
        if column not in python_columns
    ]

    print("Columns missing from SQL:", missing_in_sql)
    print("Columns missing from Python:", missing_in_python)

    raise ValueError(
        "Pipeline stopped because the Python and SQL columns "
        "do not match exactly."
    )

print("Validation passed: Python and SQL columns match.")
```

```python
expected_rows = 2005
actual_rows = len(df_clean)
missing_values = int(df_clean.isnull().sum().sum())
duplicate_rows = int(df_clean.duplicated().sum())

print("Rows ready for loading:", actual_rows)
print("Missing values:", missing_values)
print("Duplicate rows:", duplicate_rows)

if actual_rows != expected_rows:
    raise ValueError(
        f"Pipeline stopped: expected {expected_rows} rows, "
        f"but found {actual_rows}."
    )

if missing_values > 0:
    raise ValueError(
        "Pipeline stopped because missing values were found."
    )

if duplicate_rows > 0:
    raise ValueError(
        "Pipeline stopped because duplicate rows were found."
    )

print("Data-quality validation passed.")
```

```python
from datetime import datetime
from sqlalchemy import text

production_table = "neobank_customer_churn_cleaned"

pipeline_start = datetime.now()

try:
    # Begin one database transaction
    with engine.begin() as connection:

        # Remove the old records but preserve the table structure
        connection.execute(
            text("""
                DELETE FROM dbo.neobank_customer_churn_cleaned
            """)
        )

        # Insert the latest cleaned data into the existing table
        df_clean.to_sql(
            name=production_table,
            con=connection,
            schema="dbo",
            if_exists="append",
            index=False,
            chunksize=500
        )

        # Confirm the number of rows loaded
        loaded_rows = connection.execute(
            text("""
                SELECT COUNT(*)
                FROM dbo.neobank_customer_churn_cleaned
            """)
        ).scalar()

        if loaded_rows != len(df_clean):
            raise ValueError(
                f"Row-count mismatch: Python contains "
                f"{len(df_clean)} rows, but SQL Server contains "
                f"{loaded_rows} rows."
            )

    pipeline_end = datetime.now()
    duration = pipeline_end - pipeline_start

    print("Production pipeline completed successfully.")
    print(f"Rows loaded into SQL Server: {loaded_rows}")
    print(f"Pipeline started: {pipeline_start}")
    print(f"Pipeline completed: {pipeline_end}")
    print(f"Pipeline duration: {duration}")

except Exception as error:
    print("Pipeline failed.")
    print("The transaction was rolled back.")
    print("Error details:", error)
    raise
```
